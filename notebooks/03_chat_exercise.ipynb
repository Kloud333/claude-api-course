```python
from dotenv import load_dotenv
load_dotenv()

import sys
sys.path.append("..")
from helpers.chat_utils import add_user_message, add_assistant_message, chat

print("Helpers завантажено")
```

## Рішення (як у відео курсу)

⚠️ `input()` у Jupyter відкриває окреме текстове поле над клітинкою —
друкуй туди й тисни Enter. Цикл `while True` нескінченний: щоб
зупинити — Kernel → Interrupt (або квадратик ⏹ біля клітинки).

```python
# Список messages живе ПОЗА циклом — так історія розмови
# накопичується з кожним новим колом while
messages = []

while True:
    # Крок 1: питаємо юзера
    user_input = input("> ")
    print(">", user_input)

    # Крок 2: додаємо ввід юзера в messages
    add_user_message(messages, user_input)

    # Крок 3: викликаємо Claude через chat() — messages передається
    # ПОВНІСТЮ, з усією попередньою історією
    answer = chat(messages)

    # Крок 4: додаємо відповідь Claude в messages (щоб пам'ятав і це)
    add_assistant_message(messages, answer)

    # Крок 5: друкуємо відповідь
    print("---")
    print(answer)
    print()

    # Крок 6: while True сам собою повертає нас до кроку 1
```

## ➕ Бонус (не з умови вправи): вихід за командою

Курс навмисно лишає `while True` без виходу (зупиняти через Interrupt).
Практичніший варіант — команда `exit`, щоб не смикати kernel щоразу.

```python
messages_v2 = []

while True:
    user_input = input("> ")
    if user_input.lower() in ("exit", "quit"):
        print("Бувай!")
        break

    print(">", user_input)
    add_user_message(messages_v2, user_input)
    answer = chat(messages_v2)
    add_assistant_message(messages_v2, answer)

    print("---")
    print(answer)
    print()
```
