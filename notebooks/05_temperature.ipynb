# 05 — Temperature

Temperature (0-1) — "крутилка креативності": 0 = детерміновано,
1 = максимально креативно/варіативно. Пробуємо на генерації ідей для фільму.

```python
from dotenv import load_dotenv
load_dotenv()

import sys
sys.path.append("..")
from helpers.chat_utils import add_user_message, chat

print("Helpers завантажено (chat тепер приймає temperature=... і model_override=...)")
```

## ⚠️ Важливо: temperature НЕ працює з Claude Sonnet 5

Claude Sonnet 5 (наша дефолтна модель у `helpers/chat_utils.py`) —
як і будь-яка модель новіша за Opus 4.6 — **повністю деприкейтила**
`temperature`/`top_p`/`top_k`. Сам факт наявності поля `temperature`
у запиті (навіть `=1.0`!) кидає `400 BadRequestError`.

Ці нові моделі керують варіативністю самі, через adaptive thinking —
без зовнішньої "крутилки". Щоб реально ПОБАЧИТИ ефект temperature
(мета цього модуля курсу), використовуємо `model_override` на модель,
яка ще підтримує параметр: `claude-sonnet-4-5-20250929`.

## Low temperature — передбачувано

Запусти клітинку кілька разів (Ctrl+Enter кілька разів поспіль) —
порівняй, наскільки однакові відповіді.

```python
messages_low = []
add_user_message(messages_low, "Give me a one-sentence idea for a movie plot.")

print(chat(
    messages_low,
    temperature=0.0,
    model_override="claude-sonnet-4-5-20250929",  # temperature тут ще підтримується
))
```

## High temperature — креативно

Той самий промпт, той самий Claude — запусти кілька разів і побач
різницю у варіативності порівняно з клітинкою вище.

```python
messages_high = []
add_user_message(messages_high, "Give me a one-sentence idea for a movie plot.")

print(chat(
    messages_high,
    temperature=1.0,
    model_override="claude-sonnet-4-5-20250929",
))
```

## 🧪 Своя перевірка

Спробуй те саме питання на 3 різних temperature (0.0, 0.5, 1.0)
і порівняй усі три відповіді поруч.

```python
prompt = "Твоє питання тут"

for temp in (0.0, 0.5, 1.0):
    m = []
    add_user_message(m, prompt)
    print(f"--- temperature={temp} ---")
    print(chat(m, temperature=temp, model_override="claude-sonnet-4-5-20250929"))
    print()
```
