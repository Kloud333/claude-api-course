# 04 — System Prompts

System prompt керує **ЯК** Claude відповідає (тон/стиль/роль), не **ЩО** відповідає.

Наскрізний приклад курсу: **Math Tutor**, який має давати підказки,
а не готові відповіді.

```python
from dotenv import load_dotenv
load_dotenv()

import sys
sys.path.append("..")
from helpers.chat_utils import add_user_message, add_assistant_message, chat

print("Helpers завантажено (chat тепер приймає system=...)")
```

## Без system prompt

Claude, найімовірніше, одразу видасть повний покроковий розв'язок.

```python
messages = []
add_user_message(messages, "How do I solve 5x + 2 = 3 for x?")

answer_no_system = chat(messages)
print(answer_no_system)
```

## З math tutor system prompt

Той самий Claude, той самий алгоритм — інакша ПОВЕДІНКА, бо змінили
тільки `system`, не саме питання.

```python
# Потрійні лапки — щоб зручно писати багаторядковий текст.
# Перший рядок system prompt зазвичай задає РОЛЬ, далі — конкретні
# поведінкові інструкції (що робити / чого не робити)
system_prompt = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""

messages_tutor = []
add_user_message(messages_tutor, "How do I solve 5x + 2 = 3 for x?")

answer_with_system = chat(messages_tutor, system=system_prompt)
print(answer_with_system)
```

## 🧪 Своя перевірка

Спробуй іншу роль (напр. "суворий code reviewer", "дружній копірайтер")
на тому самому питанні й порівняй тон відповіді.

```python
my_system = """
Твоя роль тут
"""

my_messages = []
add_user_message(my_messages, "Твоє питання тут")
print(chat(my_messages, system=my_system))
```

## ➕ System Prompts Exercise (з відео)

System prompt працює не тільки для тону в чаті — так само керує
**стилем виводу коду**. Порівняй звичайну відповідь і "дуже лаконічний
Python-інженер" на одне й те саме питання.

```python
# Без system prompt — очікуємо розгорнуту відповідь з поясненнями
messages_plain = []
add_user_message(
    messages_plain,
    "Write a Python function that checks a string for duplicate characters.",
)

print(chat(messages_plain))
```

```python
# З system prompt — просимо конкретний, лаконічний стиль
messages_concise = []
add_user_message(
    messages_concise,
    "Write a Python function that checks a string for duplicate characters.",
)

answer = chat(
    messages_concise,
    system="You are a Python engineer who writes very concise code",
)
print(answer)
```
